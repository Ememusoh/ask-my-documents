# Test Retriever

This notebook tests `rag/retriever.py`.

Pinecone does the vector search. The retriever filters results, checks the relevance threshold, and prepares citations.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "rag").exists():
    project_root = project_root.parent

sys.path.insert(0, str(project_root))
print(f"Project root: {project_root}")
```

## 1. Test threshold filtering with fake Pinecone results

This local test does not call Pinecone.

```python
from rag.retriever import filter_by_relevance, prepare_retrieved_chunk

fake_results = [
    {
        "score": 0.91,
        "chunk_text": "This is strong supporting evidence.",
        "document_name": "demo.txt",
        "source_label": "Paragraph 1",
    },
    {
        "score": 0.30,
        "chunk_text": "This is weak evidence.",
        "document_name": "demo.txt",
        "source_label": "Paragraph 2",
    },
]

relevant_results = filter_by_relevance(fake_results, relevance_threshold=0.5)
clean_chunks = [prepare_retrieved_chunk(result) for result in relevant_results]

clean_chunks
```
