# Test Prompts

This notebook tests `rag/prompts.py`.

The prompt tells the model to answer only from retrieved document passages and cite sources.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "rag").exists():
    project_root = project_root.parent

sys.path.insert(0, str(project_root))
print(f"Project root: {project_root}")
```

## 1. Build a grounded prompt from fake retrieved chunks

This local test does not call OpenAI or Pinecone.

```python
from rag.prompts import build_grounded_prompt, build_chat_messages

question = "What is the goal of the project?"

retrieved_chunks = [
    {
        "text": "Ask My Documents is a Retrieval-Augmented Generation prototype for answering questions from uploaded documents.",
        "citation": "Project_Proposal.docx, Paragraph 2",
        "score": 0.87,
    },
    {
        "text": "The system should show supporting source passages so answers can be checked.",
        "citation": "Project_Proposal.docx, Paragraph 4",
        "score": 0.76,
    },
]

prompt = build_grounded_prompt(question, retrieved_chunks)
print(prompt)
```

## 2. Build chat messages for the future generator

This is the format `rag/generator.py` can send to the OpenAI API later.

```python
messages = build_chat_messages(question, retrieved_chunks)
messages
```

## 3. Test the abstention prompt

When there are no retrieved chunks, the prompt tells the model to abstain.

```python
empty_prompt = build_grounded_prompt(
    question="What is the author's favorite food?",
    retrieved_chunks=[],
)

print(empty_prompt)
```
